# JSON (JavaScript Object Notation) 

- is a lightweight data format used to store and exchange data between applications.

# It is commonly used in:

- Web APIs

- Web applications

- Data storage

- Configuration files

- JSON is easy for humans to read and easy for machines to parse.

```python
{
  "name": "Suraj",
  "age": 22,
  "city": "Delhi",
  "skills": ["Python", "Power BI", "SQL"]
}
```

# JSON structure uses:

- Curly brackets {} → Object

- Square brackets [] → Array / List

- Key : Value pairs

# 2️⃣ How to Load JSON Data (Python)

- In Python we use the json library.

- import json

# 3️⃣ Load JSON from String

```python
import json

data = '{"name":"Suraj","age":22,"city":"Delhi"}'

result = json.loads(data)         # 👉 loads() converts JSON string → Python dictionary

print(result)
print(result["name"])
```

# 5️⃣ Convert Python Data to JSON

```python
import json

data = {
    "name": "Suraj",
    "age": 22,
    "city": "Delhi"
}

json_data = json.dumps(data)                # 👉 dumps() converts Python → JSON string

print(json_data)
```

# File handling with json

```python
d1 = {
    "S003": {
        "id": "Q02",
        "name": "suraj",
        "age": 22,
        "email": "s@gmail.com",
        "subject": "python",
    "city":"delhi"}
}
```

```python
import json
with open('new.json', 'w') as f:
    data = json.dumps(d1)
    f.write(data)
```

```python
d11 = {
    "S003": {
        "id": "Q01",
        "name": "suyash",
        "age": 23,
        "email": "su@gmail.com",
        "subject": "python",
    "city":"noida"}
}
```

# File handling with byte

```python
import pickle

list1 = [10, 20, 30]
enc_data = pickle.dumps(list1)
print(enc_data, type(enc_data))
with open('bytes_file.py', 'wb') as f:
    f.write(enc_data)
    py_data = pickle.loads(enc_data)
    print(py_data,type(py_data))
```
